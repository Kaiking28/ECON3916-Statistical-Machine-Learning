```python
# Step 1
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from statsmodels.tools.eval_measures import rmse
import matplotlib.pyplot as plt

# Load dataset
url = '/content/Zillow_ZHVI_2026_Micro.csv'
df = pd.read_csv(url)

# Quick EDA
print(df.head())
print(df.describe())


# Step 2
formula = 'Home_Value ~ Square_Footage + Property_Age + Distance_to_Transit'


# Step 3
model = smf.ols(formula=formula, data=df)
results = model.fit()
print(results.summary())


# Step 4
y_pred = results.predict(df)


# Step 5
model_rmse = rmse(df['Home_Value'], y_pred)
print(f"\nThe Predictive RMSE is: ${model_rmse:,.2f}")
```
